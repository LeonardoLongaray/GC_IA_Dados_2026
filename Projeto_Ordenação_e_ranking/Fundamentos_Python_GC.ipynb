# Sexta função (Extra)
##### Enumerate(Criar um índice para ordenar)

```python
# Utilizando Sorted e Enumerate para criar uma lista ordenada
# por número de acordo com a nota dos filmes
ranking = sorted(slasher, key=lambda filme: filme["nota"], reverse=True)

# Cria o Ranking como uma variável válida pra print depois

for posicao, filme in enumerate(ranking, start=1):
    print(f"{posicao}º - {filme['filme']} | Nota: {filme['nota']}")

# A estrutura com "|" já deixa tabulada para csv
```
