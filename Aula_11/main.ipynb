```python
import random # Importando biblioteca para randomizar

# Dicionário com os sabores e preços
cardapio = {
    "chocolate": 5.00,
    "baunilha": 4.50,
    "morango": 3.00,
    "flocos": 9.00,
    "menta": 6.00,
    "abacaxi": 7.00
}

# Dicionário com as cores para cada sabor
#  Código         Cores
# \033[91m	🔴 Vermelho claro
# \033[31m	🔴 Vermelho
# \033[92m	🟢 Verde claro
# \033[93m	🟡 Amarelo
# \033[94m	🔵 Azul
# \033[95m	🟣 Magenta
# \033[96m	🩵 Ciano
# \033[0m	⚪ Volta ao normal
cores = {'chocolate': '\033[90m', # Amarelo escuro == Marrom ou preto
         'baunilha': '\033[93m', # Branco
         'morango': '\033[91m', # Vermelho
         'flocos': '\033[97m', # Branco brilhante
         'menta': '\033[92m', # Verde
         'abacaxi': '\033[95m'} # Amarelo

# Lista de brindes disponíveis
brindes = ["Canudo", "Copo personalizado", "Gelo", "Badge"]

# Função para mostrar o cardápio
def mostrar_cardapio():
    print('\033[91m''-- CARDAPIO --''\033[0m')
    for sabor, preco in cardapio.items():
        cor = cores[sabor]
        print(f"{cor}{sabor}, R${preco:.2f}\033[0m")

# Função para fazer o pedido
def fazer_pedido():
    total = 0
    pedido = []
    while True:
        sabor = input("\nEscolha o sabor: (digite 'fechar' para sair)")
        if sabor == 'fechar':
            break
        elif sabor in cardapio:
            total += cardapio[sabor]
            pedido.append(sabor)
            cor = cores[sabor]
            print(f"{sabor} adicionado!")
        else:
            print("Sabor não está disponível no cardápio.")
    return pedido, total

mostrar_cardapio()
pedido, total = fazer_pedido()

print(f'\nSeu pedido: {pedido}')
print(f'Total: R${total:.2f}')

if total> 15:
    print(f'\033[92mVocê ganhou um brinde: {random.choice(brindes)}''\033[0m')
```
