```python
from flask import Flask, app, jsonify
import requests

pokedex = []
pedido = requests.get('https://pokeapi.co/api/v2/pokemon?limit=50')

@app.route('/listartodos', methods=['GET'])
def listar_dex():
    for i in pedido:
        pokedex.append(i)
    return jsonify(pokedex)
@app.route('/listarpokemon/<int:id>', methods=['GET'])
def listar_pokemon(id):
    for i in pedido:
        if i['id'] == id:
            return jsonify(i)
    return jsonify({'error': 'Pokemon not found'}), 404


```
